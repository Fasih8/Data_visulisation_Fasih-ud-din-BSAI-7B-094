# Task 4 — Choose the right chart five times

**Week 1 data visualization assignment**

This notebook follows the task brief and chart-design rules: titles state findings, bars use a zero baseline, color carries meaning, excluded rows are stated, and each chart has a short rationale.

## Setup and data

Run the cells from top to bottom. The CSV is read from its public source URL; an internet connection is required. If needed, install `pandas`, `numpy`, and `matplotlib` in your notebook environment.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
DATA_URL = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv'
df = pd.read_csv(DATA_URL)
df.head()
```

## Data quality

Horsepower is missing for some cars. The five questions below do not require horsepower, so retain all available rows for those charts and report the missing count. Only exclude rows if a plotted variable is missing; state that count on the chart.

```python
print('Rows:',len(df),'| Missing values by column:\n',df.isna().sum())
print('Cars with missing horsepower:',int(df.horsepower.isna().sum()))
```

## 1. Did cars become more fuel-efficient?

```python
yearly=df.groupby('model_year').mpg.mean()
fig,ax=plt.subplots(figsize=(8,4.5)); ax.plot(1900+yearly.index,yearly.values,marker='o',color='#0072B2')
ax.set(title='Average fuel economy increased across model years',xlabel='Model year',ylabel='Mean fuel economy (mpg)'); fig.tight_layout()
```

**Why a line chart:** model year is ordered, and the line makes the long-term change in the mean easy to see.

## 2. Do origins differ in fuel efficiency?

```python
origin_order=['usa','europe','japan']
fig,ax=plt.subplots(figsize=(8,5))
data=[df.loc[df.origin.eq(o),'mpg'] for o in origin_order]
ax.boxplot(data,tick_labels=[o.title() for o in origin_order],showfliers=False)
ax.set(title='Japanese and European cars generally achieve higher fuel economy',xlabel='Origin',ylabel='Fuel economy (mpg)'); fig.tight_layout()
```

**Why a box plot:** it compares the distributions and medians by origin, showing spread as well as typical fuel economy.

## 3. Weight and fuel economy

```python
fig,ax=plt.subplots(figsize=(8,5)); ax.scatter(df.weight,df.mpg,alpha=.65,s=24,color='#0072B2')
ax.set(title='Heavier cars tend to use more fuel per mile',xlabel='Weight (lbs)',ylabel='Fuel economy (mpg)'); fig.tight_layout()
```

**Why a scatter plot:** both variables are numeric, and each point shows a car, making the relationship and any outliers visible.

## 4. Distribution of cylinder counts

```python
counts=df.cylinders.value_counts().sort_index()
fig,ax=plt.subplots(figsize=(7,4.5)); ax.bar(counts.index.astype(str),counts.values,color='#7A9E7E')
ax.set(title='Eight-cylinder cars are most common in this fleet',xlabel='Cylinders',ylabel='Number of cars'); ax.set_ylim(bottom=0); fig.tight_layout()
```

**Why a bar chart:** cylinder count is a small set of discrete categories, and bars make their frequencies easy to compare with a zero baseline.

## 5. Did cars get lighter as fuel economy improved?

```python
weight_year=df.groupby('model_year').weight.mean()
fig,ax=plt.subplots(figsize=(8,4.5)); ax.plot(1900+weight_year.index,weight_year.values,marker='o',color='#D55E00')
ax.set(title='Average vehicle weight fell over the same model years',xlabel='Model year',ylabel='Mean weight (lbs)'); fig.tight_layout()
```

**Why a line chart:** it tracks the yearly mean weight over the same ordered axis as fuel economy. Together the two time trends are consistent with lighter cars contributing to improved mpg, though these aggregate trends alone do not prove causation.

## Redraw the weakest chart

The origin comparison can hide the shape and overlap of the distributions behind three summary boxes. Redraw it with a violin plot, while retaining median markers for a direct comparison.

```python
fig,ax=plt.subplots(figsize=(8,5)); parts=ax.violinplot(data,showmedians=True,showextrema=False)
for body in parts['bodies']: body.set_facecolor('#56B4E9'); body.set_edgecolor('white'); body.set_alpha(.8)
ax.set_xticks([1,2,3], [o.title() for o in origin_order])
ax.set(title='Fuel economy distributions vary by origin, with different shapes',xlabel='Origin',ylabel='Fuel economy (mpg)'); fig.tight_layout()
```

**What changed:** the violin chart shows the distribution shape and concentration as well as a median marker, which the box plot compresses. It still compares the same variable by origin and keeps the reading task focused.
