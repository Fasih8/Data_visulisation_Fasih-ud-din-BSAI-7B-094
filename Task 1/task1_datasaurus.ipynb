## Setup and data

Run the cells from top to bottom. The CSV is read from its public source URL; an internet connection is required. If needed, install `pandas`, `numpy`, and `matplotlib` in your notebook environment.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
DATA_URL = 'https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-10-13/datasaurus.csv'
df = pd.read_csv(DATA_URL)
df.head()
```

## Summary statistics for all 13 datasets

Compute each dataset's means, sample standard deviations, and Pearson correlation in one table.

```python
summary = df.groupby('dataset').agg(mean_x=('x','mean'), mean_y=('y','mean'), sd_x=('x','std'), sd_y=('y','std'))
summary['correlation'] = pd.Series({name: part.x.corr(part.y) for name, part in df.groupby('dataset')})
summary.round(3)
```

### What the table alone might suggest

If these numbers were all I had, I would conclude that the 13 groups have essentially the same centers, spreads, and linear association, so their distributions and relationships are probably similar. That conclusion would be unjustified: summary statistics describe only selected numeric features, not the shape or structure of the points.

## The 13 scatter plots

```python
datasets = sorted(df['dataset'].unique())
fig, axes = plt.subplots(4, 4, figsize=(14, 11), sharex=True, sharey=True)
for ax, name in zip(axes.flat, datasets):
    part = df[df['dataset'] == name]
    ax.scatter(part['x'], part['y'], s=13, alpha=.75, color='#2878B5')
    ax.set_title(name)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
for ax in axes.flat[len(datasets):]: ax.set_visible(False)
fig.suptitle('The same summary statistics conceal 13 very different shapes', fontsize=16)
fig.tight_layout()
```

### What the plots show

The scatter plots reveal distinct forms: curved and linear patterns, clusters, a nearly circular cloud, and recognizable structures including a dinosaur. Thus having 13 groups and 142 observations per group does not make the summary statistics sufficient. Means, standard deviations, and correlation remain useful summaries, but they cannot replace looking at the data. The plots expose structure, outliers, and nonlinear relationships those summaries hide.
